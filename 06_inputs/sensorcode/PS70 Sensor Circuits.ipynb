```python
# Import Libraries:
import numpy as np                  # Numerical Python
import matplotlib.pyplot as plt     # Matplotlib for visualization
import pandas as pd                 # Data analysis
```

```python
file_path1 = 'Sensor_Value_Circuits_Sheet1.csv'
LED_data = pd.read_csv(file_path1)

length = LED_data['Distance']
sensorVal = LED_data['SensorValue']

plt.plot(length, sensorVal, marker='o')
plt.xlabel('Distance of light source from sensor (mm)')
plt.ylabel('Sensor Value')
plt.title('Distance from Light Sensor vs Sensor Value')
plt.grid()
plt.show()
```

```python
file_path2 = 'Sensor_Value_Circuits_Sheet2.csv'
CP_data = pd.read_csv(file_path2)

length = CP_data['Degrees']
sensorVal = CP_data['Sensor Value']

plt.plot(length, sensorVal, marker='o')
plt.xlabel('Rotation of Capacitor (degrees)')
plt.ylabel('Sensor Value')
plt.title('Rotation Angle vs Sensor Value')
plt.grid()
plt.show()
```
